```python
import cv2
import matplotlib.pyplot as plt
from clowner import Clowner
from face_recognitio import FaceRecognizer
import numpy as np
```

```python
face_db=FaceRecognizer()
face_db.build_database("./face_base/")

clowner =Clowner()
```

```python
# face_db.display_database_info()

img=cv2.imread('')
results=face_db.search(img)

def expand_bb(bb):
    x, y, x1, y1 = bb
    w = x1 - x
    h = y1 - y
    pad_w = w // 3
    pad_h = h // 3
    return (x - pad_w, y - pad_h, x1 + pad_w, y1 + pad_h)

for id,bb,similarity in results:
    name=face_db.get_name_by_index(id)
    print(name,similarity)
    if name=="danya":
        if similarity > 0.1:
            bb=expand_bb(bb)
            x, y, x1, y1 = list(map(int,bb))
            face_crop = img[y:y1, x:x1,:]
            new_crop=clowner.process(np.array(face_crop))
            img[y:y1, x:x1,:]=new_crop
            
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```
